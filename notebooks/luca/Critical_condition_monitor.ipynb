```python

import time
import random
def monitor_analysis(altitude):
    # in caso d svenimento printa (stato, sintomi, tempo di coscienza utile)
    if altitude < 1500:
        return "NOMINALE", "Nessuno (Condizioni di sicurezza)", "soglia di sicurezza"
        
    elif 1500 <= altitude < 4500:
        return "PRE-ALLERTA", "Riduzione della visione notturna", "Più di 10 minuti"
        
    elif 4500 <= altitude < 6000:
        return "ATTENZIONE", "Dispnea da sforzo (Respiro corto e faticoso)", "Circa 4-5 minuti"
        
    elif 6000 <= altitude < 7500:
        return "PERICOLO", "Perdita di realtà, vertigini, formicolii", "Meno di 3 minuti"
        
    else:
        return "CRITICO ELEVATO", "Perdita di conoscenza imminente!", "Pochi secondi (9-15s)"

# --- SIMULAZIONE DEL MONITOR IN TEMPO REALE ---
print("=== SISTEMA MEGATRON: Monitoraggio Ipossia Spaziale Attivo ===")
print("Controllo della pressione e dell'altitudine della tuta in corso...\n")

try:
    while True:
        # Simuliamo un'altitudine che cambia per vedere come reagisce il monitor.
        # Nello spazio reale, un aumento di quota equivale a una perdita di pressione nella tuta.
        quota_simulata = random.randint(0, 9000)
        
        stato, sintomi, tempo_limite = monitor_analysis(quota_simulata)
        
        print(f"[Sensore Tuta] Altitudine equivalente: {quota_simulata} metri")
        print(f"Stato: {stato}")
        print(f"Sintomi rilevati: {sintomi}")
        print(f"Tempo utile per agire: {tempo_limite}")
        
        # Se lo stato è Pericolo o Critico, invia la richiesta di aiuto all'altro astronauta
        if stato in ["PERICOLO", "CRITICO ELEVATO"]:
            print("\n" + "!" * 60)
            print("ALERT EMERGENCY: L'astronauta sta perdendo conoscenza!")
            print("RICHIESTA ASSISTENZA: Intervento immediato del compagno di equipaggio!")
            print("!" * 60 + "\n")
        else:
            print("Situazione sotto controllo.\n")
        
        # Aspetta 3 secondi prima della prossima lettura dei sensori
        time.sleep(3)

except KeyboardInterrupt:
    print("\nMonitoraggio interrotto dal sistema.")


```

```python
altitude_feet=8000
oxygenperc= 20
def miscela_necessaria_ossigeno(altitude_feet, oxygenperc):
    # controlla quanto ossigeno la tuta EVA deve fornire in base all'altitudine durante il decollo
    if altitude_feet <= 10000:
        o2minimo= 27  # valore lievemente piu alto del minimo sindacale
    elif 10000 < altitude_feet <= 20000:
        o2minimo= 35
    elif 20000 < altitude_feet <= 30000:
        o2minimo= 44
    elif 30000 < altitude_feet <= 40000:
        o2minimo= 78
    elif altitude_feet > 40000:
        o2minimo= 100
    if oxygenperc < o2minimo:
        return "CRITICAL OXYGEN DEFICIENCY-Action: Switch to 100% emergency O2 supply and check suit pressure seal." 
    else:
        return "CONDITIONS: STABLE"

print(miscela_necessaria_ossigeno(altitude_feet,oxygenperc))
```

```python
def controlla_rischio_barotrauma(pressione_cabina_iniziale, pressione_cabina_attuale, tempo_trascorso_secondi):
    """
    Calcola la velocità di depressurizzazione per prevenire il barotrauma polmonare
    basandosi sul grafico della pressione transmurale.
    """
    # Calcoliamo la variazione di pressione (mmHg)
    delta_pressione = pressione_cabina_iniziale - pressione_cabina_attuale
    
    # Calcoliamo la velocità di crollo (mmHg al secondo)
    if tempo_trascorso_secondi > 0:
        velocita_depressurizzazione = delta_pressione / tempo_trascorso_secondi
    else:
        velocita_depressurizzazione = 0

    # SOGLIA CRITICA: Se la pressione crolla di oltre 150 mmHg in meno di mezzo secondo (0.5s)
    # la pressione transmurale polmonare supera il livello di guardia (picco a 0.2s nel grafico)
    if tempo_trascorso_secondi <= 0.5 and delta_pressione > 150:
        return "CRITICAL EMERGENCY: DEPRESSURIZZAZIONE ESPLOSIVA!", "Rischio imminente di Barotrauma Polmonare (lacerazione alveolare). Espellere aria dai polmoni!"
    
    elif velocita_depressurizzazione > 50:
        return "WARNING: DEPRESSURIZZAZIONE RAPIDA", "Perdita di pressione controllata ma pericolosa. Monitorare pressione polmonare."
    
    else:
        return "STATUS: NOMINAL", "Pressione della cabina stabile o variazioni nei limiti di sicurezza."

# --- TEST PER IL TUO MONITOR MEGATRON ---
# Immaginiamo che la cabina passi da 350 mmHg a 50 mmHg in soli 0.2 secondi (proprio come nel grafico!)
pressione_start = 350
pressione_now = 50
tempo = 0.2

stato_allarme, indicazioni = controlla_rischio_barotrauma(pressione_start, pressione_now, tempo)

print(f"Stato Monitor: {stato_allarme}")
print(f"Indicazioni per l'equipaggio: {indicazioni}")
```

```python
def analizza_strato_atmosferico(altitude_feet):
   # Rileva lo strato atmosferico e i rischi associati (Temperatura e Radiazioni UV)
    #basandosi sul grafico della Troposfera/Stratosfera/Mesosfera.
    
    if altitude_feet < 36000:
        strato = "TROPOSFERA"
        temp_esterna_stimata = 15 - (altitude_feet / 500)  # Calo termico lineare standard
        rischio_radiazioni = "Basso (Protetto dall'intera atmosfera)"
        indicazione_tuta = "Riscaldamento standard attivato."
        
    elif 36000 <= altitude_feet < 160000:
        strato = "STRATOSFERA"
        temp_esterna_stimata = -55 if altitude_feet < 65000 else 0  # Inversione termica per ozono
        rischio_radiazioni = "Moderato (Presenza dello scudo di Ozono)"
        indicazione_tuta = "Schermatura UV nominale. Regolazione termica attiva."
        
    elif 160000 <= altitude_feet < 280000:
        strato = "MESOSFERA"
        temp_esterna_stimata = -90  # Il punto più freddo dell'atmosfera nel grafico
        rischio_radiazioni = "Elevato (Sopra lo scudo di Ozono)"
        indicazione_tuta = "ATTENZIONE: Temperature esterne estremamente rigide. Attivare massimo riscaldamento energetico."
        
    else:
        # Termosfera ed Esosfera (Spazio Aperto)
        strato = "TERMOSFERA / SPAZIO APERTO"
        temp_esterna_stimata = 500  # Esposizione solare diretta senza filtro
        rischio_radiazioni = "CRITICO (Esposizione totale a raggi UV e cosmici)"
        indicazione_tuta = "CRITICAL: Attivare deflettore radiativo e sistema di refrigerazione EVA!"

    return strato, temp_esterna_stimata, rischio_radiazioni, indicazione_tuta

# --- TEST DEL MODULO ATMOSFERICO SUL MONITOR ---
# Ipotizziamo che l'astronauta sia in fase di decollo o passeggiata spaziale a 200.000 piedi (Mesosfera)
quota_attuale = 200000

strato, temp_ext, radiazioni, azione_megatron = analizza_strato_atmosferico(quota_attuale)

print(f"=== MONITOR ATMOSFERICO MEGATRON ===")
print(f"Posizione attuale: {strato}")
print(f"Temperatura esterna stimata: {temp_ext}°C")
print(f"Livello Radiazioni Esterne: {radiazioni}")
print(f"Direttiva Tuta Sensori: {azione_megatron}")
```

```python
# presenza di persone che si allenano alterano la pressione
def monitora_atmosfera_stazione_avanzato(
    num_astronauti, 
    volume_modulo_m3, 
    o2_perc_attuale, 
    co2_perc_attuale,
    pressione_attuale_mmhg, 
    in_allenamento=False,
    eclss_attivo=True
):
   #modello ECLSS
    volume_litri = volume_modulo_m3 * 1000
    
    # 1. Consumo O2 e produzione CO2 per astronauta (Quoziente Respiratorio RQ ~ 0.85)
    consumo_o2_persona = 1.5 if in_allenamento else 0.3      # Litri/minuto
    produzione_co2_persona = consumo_o2_persona * 0.85       # Litri/minuto
    
    consumo_o2_ora = num_astronauti * consumo_o2_persona * 60
    produzione_co2_ora = num_astronauti * produzione_co2_persona * 60
    
    # 2. Compensazione del sistema di bordo (OGS per O2, CDRA per CO2)
    # Un sistema nominale compensa fino a 6 astronauti a riposo (~108 L/h di O2)
    reintegro_o2_ora = 110.0 if eclss_attivo else 0.0
    rimozione_co2_ora = 95.0 if eclss_attivo else 0.0
    
    # 3. Bilancio netto dei gas in 1 ora
    bilancio_o2_litri = reintegro_o2_ora - consumo_o2_ora
    bilancio_co2_litri = produzione_co2_ora - rimozione_co2_ora
    
    o2_stimato_1h = round(o2_perc_attuale + (bilancio_o2_litri / volume_litri) * 100, 2)
    co2_stimata_1h = round(max(0.04, co2_perc_attuale + (bilancio_co2_litri / volume_litri) * 100), 3)
    
    # 4. Pressione parziale dell'ossigeno (Legge di Dalton)
    ppo2_attuale = round(pressione_attuale_mmhg * (o2_perc_attuale / 100), 1)
    
    # 5. Valutazione clinica e ambientale (Standard NASA)
    allarmi = []
    if o2_perc_attuale < 19.5 or ppo2_attuale < 148:
        allarmi.append("CRITICAL: Ipossia ambientale (O2 basso)")
    if o2_perc_attuale > 23.5:
        allarmi.append("CRITICAL: Rischio incendio (O2 troppo alto)")
    if co2_perc_attuale > 0.5:  # Soglia NASA di attenzione per performance cognitive (~3.8 mmHg)
        allarmi.append("WARNING: Ipercapnia (CO2 elevata - calo cognitivo)")
    if o2_stimato_1h < 19.5 or co2_stimata_1h > 0.5:
        allarmi.append("ALERT PREVENTIVO: Sovraffollamento modulo rispetto alla capacità ECLSS")
        
    stato = "NOMINAL" if not allarmi else " | ".join(allarmi)

    return {
        "stato": stato,
        "ppo2_mmhg": ppo2_attuale,
        "o2_proiezione_1h": o2_stimato_1h,
        "co2_proiezione_1h": co2_stimata_1h,
        "bilancio_netto_o2_L_h": round(bilancio_o2_litri, 1)
    }
```
